```python
import pandas as pd
df = pd.read_csv("LMS_Dirty_Dataset.csv")
print(df)
print(df.head(10))
print(df.tail(10))
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.describe())
```

```python
df = df.drop_duplicates()
print("Duplicates after removing:", df.duplicated().sum())
print(df.isnull().sum())
numeric_cols = df.select_dtypes(include=np.number).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
categorical_cols = df.select_dtypes(include="object").columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
print("Missing Values After Cleaning")
print(df.isnull().sum())
df["enrollment_date"] = pd.to_datetime(df["enrollment_date"], errors="coerce")
text_cols = ["student_name","course","city","gender","payment_status","mentor","completion_status"]
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
num_cols = ["course_fee","attendance_percent","assignment_score","final_score"]
for col in num_cols:
    df.loc[df[col] < 0, col] = np.nan
    df[col] = df[col].fillna(df[col].median())
print("Data Cleaning Completed")
```

```python
cols = ["course_fee",
        "attendance_percent",
        "assignment_score",
        "final_score"]

for col in cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR


    print( col)
    print(Q1)
    print( Q3)
    print(IQR)
    print(lower)
    print(upper)

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("Number of Outliers :", len(outliers))

    print(outliers)
```

```python
df_remove = df.copy()

for col in cols:

    Q1 = df_remove[col].quantile(0.25)
    Q3 = df_remove[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_remove = df_remove[
        (df_remove[col] >= lower) &
        (df_remove[col] <= upper)
    ]

print(df_remove.shape)
```

```python
df_median = df.copy()

for col in cols:

    Q1 = df_median[col].quantile(0.25)
    Q3 = df_median[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    median = df_median[col].median()

    df_median.loc[df_median[col] < lower, col] = median
    df_median.loc[df_median[col] > upper, col] = median

print(df_median.head())
```

```python
df_clip = df.copy()

for col in cols:

    Q1 = df_clip[col].quantile(0.25)
    Q3 = df_clip[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clip[col] = df_clip[col].clip(lower, upper)

print(df_clip.head())
```

```python
from scipy.stats import zscore

z_df = df.copy()

for col in cols:

    z_df[col + "_zscore"] = zscore(z_df[col])

print(z_df.head())
```
