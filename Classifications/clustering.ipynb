```python
import pandas as pd
import numpy as np 
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("clusteringdata.csv")
# print(df.head())
# print(df.shape)
# print(df.describe())

# print(df.isnull().sum())
# print(df.duplicated().sum())

#yeti tapai ko kunai columns ma null value xa vane or missing values 
numerical_columns = [
    "StudyHoursPerWeek", 
    "AttendancePercentage",
    "AssignmentCompletionPercentage"
]
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

#feature selection 
X = df[
    [
    "StudyHoursPerWeek", 
    "AttendancePercentage",
    "AssignmentCompletionPercentage"
    ]
]

#standardScalar Object create 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
#make model 
kmeans = KMeans(
    n_clusters = 3, #kati ota clusters create garne (excellent, good, poor)
    init="k-means++", #starting centroid choose garne based on how well spread start 
    n_init = 10, #run above init or k means ++  10 times with differnt initial centroid choices
    random_state = 42, 
    tol=0.0001 #kati khera chai entire hamro model interation process stop garne tw 
)
df["Group"] = kmeans.fit_predict(X_scaled)
```
