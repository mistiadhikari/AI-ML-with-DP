```python
import pandas as pd

from sklearn.tree import DecisionTreeClassifier


df = pd.read_csv("student_binary_classification.csv")

df["Result"] = df["Result"].map({
    "Fail": 0,
    "Good": 1,
    "Average": 2,
    "Excellent": 3
})

X = df[["StudyHours"]]
y = df["Result"]


model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)


model.fit(X, y)


print(model.tree_.threshold)

```

```python
new_students = pd.DataFrame({
    "StudyHours": [2, 3, 5]
})


predictions = model.predict(new_students)

print(predictions)
```
